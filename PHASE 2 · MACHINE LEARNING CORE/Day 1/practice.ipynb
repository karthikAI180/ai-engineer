```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([1,2,3,4,5],dtype=float)
y=np.array([2,4,5,4,5],dtype=float)
```

Implement linear regression from scratch in NumPy —
no sklearn

```python
xmean=x.mean()
ymean=y.mean()
```

Calculating Slope and Intercept

```python
numerator=0
denominator=0
for i in range(len(x)):
    numerator+=(x[i]-xmean)*(y[i]-ymean)
    denominator+=(x[i]-xmean)**2
Slope=numerator/denominator
Intercept=ymean-(Slope*xmean)
print(f"Slope of the line is {Slope}")
print(f"Intercept for the line is {Intercept}")
```

NumPy automatically applies the multiplication and addition to every element in X. That is called vectorisation — no loop needed. Where Y_hat is predicted y for each x

```python
y_hat = Slope * x + Intercept
print(y_hat)
# for i in x:
#     print(f"for the value x={i} then the y value will be {Slope*i+Intercept}")
#     y_hat.append
```

Compute MSE 

```python

Least_Squared_array=(y_hat-y)**2
MSE=Least_Squared_array.sum()/len(y_hat)
print(MSE)
```

Plot: data points, your fitted line, residuals

```python
for i in range(len(x)):
    plt.plot([x[i], x[i]], [y[i], y_hat[i]])
plt.scatter(x,y)
plt.plot(x,y_hat)
plt.xlabel('X_values')
plt.ylabel('Y_values')
plt.title("Linear regression")
plt.show()
```

Doubling y doubles slope and intercept

```python
new_y=y*2
new_ymean=new_y.mean()
new_numerator=0
for i in range(len(x)):
    new_numerator+=(x[i]-xmean)*(new_y[i]-new_ymean)
new_Slope=new_numerator/denominator
new_Intercept=new_ymean-(new_Slope*xmean)
print(f"Slope of the line is {Slope}")
print(f"Intercept for the line is {Intercept}")
print(f"Slope of the line is {new_Slope}")
print(f"Intercept for the line is {new_Intercept}")

```
