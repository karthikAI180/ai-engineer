```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
help(stats.norm.rvs)
```

Task 1: Generate 1000 random height values from a normal distribution (mean=170, std=10) and plot a histogram.

```python
normal_distribution_data=stats.norm.rvs(loc=170,scale=10,size=1000)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=normal_distribution_data,
    bins=30,
    kde=True
)

plt.title("Normal Distribution Using Seaborn")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

Task 2: Generate 1000 random values from a Poisson distribution with average = 5 and plot a histogram.

```python
poisson_distribution_data=stats.poisson.rvs(mu=5,size=1000)
plt.figure(figsize=(8, 5))

sns.histplot(
    data=poisson_distribution_data,
    bins=20,
    kde=False
)

plt.title("Poisson Distribution Using Seaborn")
plt.xlabel("Event Count")
plt.ylabel("Frequency")
plt.show()
```

Task 3:compute mean, variance, std on your Task 1 data using numpy

```python
# Mean=normal_distribution_data.mean()
# standard_deviation=normal_distribution_data.std(ddof=1)
# variance=standard_deviation**2
Mean=np.mean(normal_distribution_data)
standard_deviation=np.std(normal_distribution_data,ddof=1)
variance=np.var(normal_distribution_data,ddof=1)

print(f"The mean of the normal distribution data is {Mean}")
print(f"The variance of the normal distribution data is {variance}")
print(f"The standard_deviation of the normal distribution data is {standard_deviation}")
```
